```python
import numpy as np

def max_drawdown(returns):
    returns = np.asarry(returns)
    # cumulative product: 
    cum_returns = np.cumprod(1+returns)
    # computes the running maximum, i.e., at each time step it stores the highest value observed so far.
    peak = np.maximum.accumulate(cum_returns)
    drawdown = (cum_returns- peak)/peak

    return np.min(drawdown)
```
